# EL ATRACTOR DE LORENZ
**1. INTRODUCCIÓN**

El sistema de Lorenz, formulado por Edward Lorenz en la década de 1960, es un conjunto de ecuaciones diferenciales que describe el comportamiento de un sistema dinámico no lineal. El sistema se define mediante las siguientes ecuaciones:

\

\begin{aligned}
\dot{x} &= \sigma (y - x) \\
\dot{y} &= r x - y - x z \\
\dot{z} &= x y - β z
\end{aligned}


con los parámetros $σ, r, β > 0$.

Estas ecuaciones son no lineales y su solución es una ecuación imposible de resolver con exactitud. Por lo tanto, aplicaremos métodos numéricos para obtener soluciones aproximadas de estas funciones. El objetivo de este trabajo es explorar la dinámica del sistema de Lorenz mediante la simulación numérica para diferentes valores del parámetro $r$, así como analizar las consecuencias de elegir distintas condiciones iniciales y longitudes de integración.

\

**2. ANÁLISIS DE LOS PUNTOS CRÍTICOS**

Las ecuaciones de Lorenz son un modelo no lineal de 3 dimensiones con ecuaciones
diferenciales de primer orden:


$$F(x,y,z) = \sigma (y - x) $$

$$G(x,y,z) = r x - y - x z $$

$$H(x,y,z)= x y - β z$$

\


Para poder describir el comportamiento del sistema, se evaluara la existencia de puntos críticos.

Es decir, $(x^*,y^*,z^*)$ es un punto crítico tal que,

$$F(x^*,y^*,z^*) = G(x^*,y^*,z^*) = H(x^*,y^*,z^*) = 0$$

Donde los 3 puntos criticos de las ecuaciones de Lorenz son :

$$  (0, 0, 0)\quad; \quad( \sqrt{β(\sigma -1)}, \sqrt{β(\sigma -1)},
(\sigma -1)) \quad y \quad ( -\sqrt{β(\sigma -1)}, -\sqrt{β(\sigma -1)},
(\sigma -1)).$$

\

3. **ESTABILIDAD DE LOS PUNTOS CRITICOS**

Para esto utilizaremos un método de linealización.

Se aplica la fórmula de Taylor para $F\quad , G \quad y \quad H$, y se asume que $(x^*,y^*,z^*)$ es un punto crítico aislado tal que $F(x^*,y^*,z^*) = G(x^*,y^*,z^*) = H(x^*,y^*,z^*) = 0$

\

Conviertiéndose el sistema, en un sistema linealizado:

\

$$\begin{bmatrix} F(x,y,z)\\G(x,y,z)\\H(x,y,z)\end{bmatrix}\approxeq \begin{bmatrix} -\sigma & \sigma & 0 \\(r-z) &-1& -x\\y&x& -β\end{bmatrix} \begin{bmatrix} x\\y\\z\end{bmatrix}$$

\

donde la matriz jacobiana resulta: $$J_{(x,y,z)} = \begin{bmatrix} -\sigma & \sigma & 0 \\(r-z) &-1& -x\\y&x& -β\end{bmatrix} $$

\

4. **CARACTERIZACIÓN DE LOS PUNTOS CRÍTICOS**

Para investigar el comportamiento de los puntos críticos de una sistema lineal se puede utilizar el metodo del autovalor-autovector:



*   PUNTO CRÍTICO (0, 0, 0)


La matriz jacobiana para el punto crítico $(0, 0, 0)$ resulta:

$$J_{(0,0,0)} = \begin{bmatrix} -\sigma & \sigma & 0 \\r &-1& 0\\0&0& -β\end{bmatrix} $$

Aplicando el metodo anterior, obtenemos:

$$λ_{2,3} = \frac{-(1+\sigma)\pm \sqrt{(1+\sigma)^2 - 4(\sigma - rβ)}}{2}$$

Para $β= 8/3, σ=10$ y para distintos valores de $r$ los autovalores de la matriz de coeficientes constantes en el punto crítico $(0,0,0)$ pertenecen a los ℝ, son diferentes y tienen signos opuestos. El punto $(0,0,0)$ es un punto silla.

\

*  PUNTO CRÍTICO $( \sqrt{β(\sigma -1)},\sqrt{β(\sigma -1)},(\sigma -1))$

La matriz jacobiana para el punto crítico resulta:

$$J_{(\sqrt{β(\sigma -1)},\sqrt{β(\sigma -1)},(\sigma -1))} = \begin{bmatrix} -\sigma & \sigma & 0 \\-1 &-1& -\sqrt{β(\sigma -1)}\\\sqrt{β(\sigma -1)}&\sqrt{β(\sigma -1)}& -β\end{bmatrix} $$

Aplicando el metodo anterior, obtenemos:

$$λ_{1} = -β$$

$$λ_{2,3} = \frac{(-1-\sigma)\pm \sqrt{(1+\sigma)^2 - 8\sigma}}{2}$$

Para $β= 8/3, σ=10$ y para distintos valores de $r$ los autovalores de la matriz de coeficientes constantes en el punto crítico es:

Para $r=10$ : el sistema es estable.


Para el resto de $r$: los autovalores son complejos indican una transición a un comportamiento más complicado y potencialmente caótico.

\

*  PUNTO CRÍTICO $( -\sqrt{β(\sigma -1)},-\sqrt{β(\sigma -1)},(\sigma -1))$

La matriz jacobiana para el punto crítico resulta:

$$J_{(-\sqrt{β(\sigma -1)},-\sqrt{β(\sigma -1)},(\sigma -1))} = \begin{bmatrix} -\sigma & \sigma & 0 \\-1 &-1& \sqrt{β(\sigma -1)}\\-\sqrt{β(\sigma -1)}&-\sqrt{β(\sigma -1)}& -β\end{bmatrix} $$

Aplicando el metodo anterior, obtenemos:

$$λ_{1} = -β$$

$$λ_{2,3} = \frac{(-1-\sigma)\pm \sqrt{(1+\sigma)^2 - 8\sigma}}{2}$$

Para $β= 8/3, σ=10$ y para distintos valores de $r$ los autovalores de la matriz de coeficientes constantes en el punto crítico es:

Para $r=10$ : el sistema es estable.

Para el resto de $r$: los autovalores son complejos indican una transición a un comportamiento más complicado y potencialmente caótico.

\

5. PLANO DE FASE

Cuando el punto inicial $(x_0,y_0,z_0)$ no es un punto crítico, entonces la trayectoria correspondiente es una superficie en el plano xyz, dentro del cual se mueve el punto $(x(t), y(t),z(t))$ conforme t incrementa.

Para esto implementamos el método de Euler.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
def Lorenz(X,Y,Z, T, r ,sigma=10, beta=8/3):
  XX = sigma * (Y-X)
  YY = r * X - Y - X * Z
  ZZ = X * Y - beta * Z
  return XX, YY, ZZ, T

def euler3(F,X0,Y0,Z0,T0,R,r):
  """Metodo de Euler para 3 variables"""
  t = np.empty((R + 1))
  x = np.empty((R + 1))
  y = np.empty((R + 1))
  z = np.empty((R + 1))
  t[0]= T0
  x[0]= X0
  y[0]= Y0
  z[0]= Z0
  for i in range (R):
    XX, YY, ZZ, T = Lorenz(x[i], y[i], z[i], t[i],r)
    inc_x = XX*h
    inc_y = YY*h
    inc_z = ZZ*h
    x[i+1]= x[i]+inc_x
    y[i+1]= y[i]+inc_y
    z[i+1]= z[i]+inc_z
    t[i+1]= t[i]+h
  return t,x,y,z

def grafica(a):
    plt.subplot(1,3,1)
    plt.title("x(t)")
    plt.plot(a[0],a[1],"lightblue")
    plt.grid(linestyle='--', linewidth=0.5)
    plt.xlabel("x")
    plt.ylabel("y")


    plt.subplot(1,3,2)
    plt.title("y (t)")
    plt.plot(a[0],a[2],"lightblue")
    plt.grid(linestyle='--', linewidth=0.5)
    plt.xlabel("x")
    plt.ylabel("z")


    plt.subplot(1,3,3)
    plt.title("x vs z")
    plt.plot(a[1],a[3],"lightblue")
    plt.grid(linestyle='--', linewidth=0.5)
    plt.xlabel("y")
    plt.ylabel("z")

    #PLOTEADO EN 3D
    fig = plt.figure(2)
    ax = fig.add_subplot(projection='3d')
    ax.plot(a[1], a[2], a[3])

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("z")
    ax.set_title("Atractor de Lorenz")
```

Para un $r = 10$:


Podemos ver que al avanzar el tiempo la función es atraída por un nodo atractor.

Al variar el punto de inicio se puede observar: $P_0= (1,1,1)$ forma una trayectoria, mientras que $P_0 = (-1,-1,-1)$ se la obtiene con sentido opuesto.


```python
#Parámetros
#Condicion inicial
x0,y0,z0 = 1,1,1
t0, tfinal = 0 , 100

h= 0.01
N=int(tfinal/h)

#Resolvemos
a=euler3(Lorenz,x0,y0,z0,t0,N,r=10)
grafica(a)
```

# Comparando con otro método de resolución: RK45

**Para un $r = 10:$**

Podemos ver que al avanzar el tiempo la función es atraída por un nodo atractor.

\

**Para un $r =  22, 24.5$:**

Dependiendo de la condicion inicial:

Tiende a ser atraido por un nodo. O la funcion es atraida primero hacia uno y luego hacia otro de los nodos atractores.

\

**Para un $r = 100,126.52,400$:**

Podemos ver que al avanzar el tiempo la función es atraída primero hacia uno y luego hacia el otro de los nodos atractores. Las soluciones oscilan en forma irregular, de manera aperiódicas, pero siempre permaneciendo en una region acotada, convergiendo a un atractor extraño.Este atractor extraño no es un punto de equilibrio, ni una órbita periódica, sino que se lo conoce como fractal. El atractor de Lorenz posee una forma de mariposa característica.

```python
#RESOLVI LA MISMA ECUACION PERO CON OTRO METODO: RK45
from scipy.integrate import solve_ivp

# Sistema de Lorenz
def lorenz(t, variables, sigma, r, b):
    x, y, z = variables
    dx_dt = sigma * (y - x)
    dy_dt = r * x - y - x * z
    dz_dt = x * y - b * z
    return [dx_dt, dy_dt, dz_dt]

# Parámetros
sigma = 10
b = 8/3

# Condiciones iniciales
condicion_inicial = [1, 1, 1]

# Tiempo de integración
t_span = (0, 100)
t_eval = np.linspace(*t_span, 10000)

# Lista de valores de r
r_values = [10, 22, 24.5, 100, 126.52, 400]

# Graficar
fig, axes = plt.subplots(len(r_values), 3, figsize=(18, 12))

for i, r in enumerate(r_values):
    # Resolver sistema de Lorenz
    sol = solve_ivp(lorenz, t_span, condicion_inicial, args=(sigma, r, b), t_eval=t_eval,method='RK45')


    # Graficar x(t), y(t), x vs z
    axes[i, 0].plot(sol.t, sol.y[0])
    axes[i, 0].set_title(f"x(t) para r = {r}")

    axes[i, 1].plot(sol.t, sol.y[1])
    axes[i, 1].set_title(f"y(t) para r = {r}")

    axes[i, 2].plot(sol.y[0], sol.y[2])
    axes[i, 2].set_title(f"x vs z para r = {r}")

plt.tight_layout()
plt.show()
```

```python
#PARA r=22 ,PRUEBO DISTINTOS PUNTOS INICIALES
r = 22

# Condiciones iniciales
condiciones_iniciales = [
    [-10, -10, -10],
    [-3, -3, -3],
    [5, 5, 5],
    [10, 10, 10],
]
# Crear figura para graficar
fig, axes = plt.subplots(len(condiciones_iniciales), 3, figsize=(15, 15))

# Resolver y graficar para cada condición inicial
for i, condicion in enumerate(condiciones_iniciales):
    # Resolver sistema de Lorenz
    sol = solve_ivp(lorenz, t_span, condicion, args=(sigma, r, b), t_eval=t_eval, method='RK45')

    # Graficar x(t)
    axes[i, 0].plot(sol.t, sol.y[0])
    axes[i, 0].set_title(f"x(t) para condición inicial {condicion}")
    axes[i, 0].set_xlabel('Tiempo')
    axes[i, 0].set_ylabel('x')

    # Graficar y(t)
    axes[i, 1].plot(sol.t, sol.y[1])
    axes[i, 1].set_title(f"y(t) para condición inicial {condicion}")
    axes[i, 1].set_xlabel('Tiempo')
    axes[i, 1].set_ylabel('y')

    # Graficar x vs z
    axes[i, 2].plot(sol.y[0], sol.y[2])
    axes[i, 2].set_title(f"x vs z para condición inicial {condicion}")
    axes[i, 2].set_xlabel('x')
    axes[i, 2].set_ylabel('z')

plt.tight_layout()
plt.show()
```

```python
#PARA r=24.5 ,PRUEBO DISTINTOS PUNTOS INICIALES
r = 24.5

# Condiciones iniciales
condiciones_iniciales = [
    [-10, -10, -10],
    [-4, -4, -4],
    [3, 3, 0],
    [5, 5, 5],
]
# Crear figura para graficar
fig, axes = plt.subplots(len(condiciones_iniciales), 3, figsize=(15, 15))

# Resolver y graficar para cada condición inicial
for i, condicion in enumerate(condiciones_iniciales):
    # Resolver sistema de Lorenz
    sol = solve_ivp(lorenz, t_span, condicion, args=(sigma, r, b), t_eval=t_eval, method='RK45')

    # Graficar x(t)
    axes[i, 0].plot(sol.t, sol.y[0])
    axes[i, 0].set_title(f"x(t) para condición inicial {condicion}")
    axes[i, 0].set_xlabel('Tiempo')
    axes[i, 0].set_ylabel('x')

    # Graficar y(t)
    axes[i, 1].plot(sol.t, sol.y[1])
    axes[i, 1].set_title(f"y(t) para condición inicial {condicion}")
    axes[i, 1].set_xlabel('Tiempo')
    axes[i, 1].set_ylabel('y')

    # Graficar x vs z
    axes[i, 2].plot(sol.y[0], sol.y[2])
    axes[i, 2].set_title(f"x vs z para condición inicial {condicion}")
    axes[i, 2].set_xlabel('x')
    axes[i, 2].set_ylabel('z')

plt.tight_layout()
plt.show()
```
