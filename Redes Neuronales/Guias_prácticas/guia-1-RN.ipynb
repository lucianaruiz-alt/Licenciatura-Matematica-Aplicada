**2)** *Conjetura de Collatz:* Implemente un loop usando la instrucción `while` calcule la iteración

\begin{equation}
n_{t+1} = f(n_t)
\end{equation}

donde $f(n) = n/2$ si $n$ es par y $f(n)=3n+1$ si $n$ es impar, partiendo de $n_0=10$. El loop debe detenerse cuando se cumple la condición $n_t=1$ para algún $t$. Que observa? Comente. Cuantas iteraciones realiza el loop?

**3)** Use `for` para iterar sobre la lista `["perro","gato","loro","caballo","llama"]`, imprimiendo los cada item en cada iteración.

```python
# 9.1)
# 1)
for i in range(11):
  if i % 2 == 0:
    print(i)
```

```python
# 2) Conjetura de Collatz con 'while'
n = 10
iteraciones = 0

print(f"Inicio: {n}")
while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    iteraciones += 1
    print(f"Paso {iteraciones}: {n}")

print(f"\nTotal de iteraciones: {iteraciones}")
```

```python
# 3) Loop 'for' para iterar sobre la lista e imprimir cada item
animales = ["perro", "gato", "loro", "caballo", "llama"]

for animal in animales:
    print(animal)
```

## **Ejercicio 10)** Definiendo y evaluando funciones

**1)** Implemente una función llamada `suma` que retorne la sume dos números. Testeela.

**2)** Implemente una función llamada `opcional` que adopte dos parámetros de entrada, y que el segundo de ellos sea opcional con un valor por defecto igual a 1. La función debe retornar la resta de ambos números. Testeela.

**3)** Implemente una función llamada `fact` que calcule y retorne el factorial de un número `n` si es que `n` es natural o zero. En caso contrario, que imprima el mensaje "El número n no es un entero no negativo." y retorne `None`. Testeela.

```python
# 10.1)
```

```python
# 1) Función suma
def suma(a, b):
    return a + b

# Testeo de suma
print("Test :")
print(suma(5, 3))    # Devuelve 8
print(suma(-2, 10))  # Devuelve 8
```

```python
# 2) Función opcional
def opcional(a, b=1):
    return a - b

# Testeo de opcional
print("\nTest :")
print(opcional(10))     # Usa b=1 por defecto -> 10 - 1 = 9
print(opcional(10, 4))  # Sobrescribe b=4 -> 10 - 4 = 6
```

```python
# 3) Función fact (factorial con validación)
def fact(n):
    # Verificamos si n es entero no negativo (natural o cero)
    if isinstance(n, int) and n >= 0:
        factorial = 1
        for i in range(1, n + 1):
            factorial *= i
        return factorial
    else:
        print(f"El número {n} no es un entero no negativo.")
        return None

# Testeo de fact
print("\nTest 10.3:")
print("Factorial de 5:", fact(5))    # Devuelve 120
print("Factorial de 0:", fact(0))    # Devuelve 1
print("Factorial de -3:", fact(-3))  # Imprime mensaje de error y devuelve None
print("Factorial de 3.5:", fact(3.5)) # Imprime mensaje de error y devuelve None
```

## **Ejercicio 12)** Practicando con clases

**1)** Defina una clase llamada `MiClase` que posea una variable miembro llamada `nombre`. El inicializador `__init__()` de la clase debe poseer un argumento opcional, que debe usarse para inicializar el valor de la variable miembro `nombre` con el valor por defecto `"mi clase"`.

**2)** Agregue a la clase la función miembro `__str__()` que retorne el valor de la variable miembro `nombre`.

**3)** Cree un objeto de la clase en cuestión y de nombre `mi_objeto`, y testee la función `__str__()` evaluando la expresión `str(mi_objeto)`.

**4)** Agregue a la clase la función `__len__()` que retorne el largo de la variable miembro `nombre`. Testee dicha función.

**5)** Agregue a la clase una función miembro llamada `cambiar_nombre` que sirva para cambiar el valor de la variable miembro `nombre`. Testee dicha función.

```python
# 11.1)
```

```python
class MiClase:
#self: referencia al propio objeto

    def __init__(self, nombre="mi clase"):
        # método constructor
        self.nombre = nombre

    def __str__(self):
        # representación de texto
        return self.nombre

    def __len__(self):
        # define la función len
        return len(self.nombre)

    def cambiar_nombre(self, nuevo_nombre):
        # Permite modificar el valor de 'nombre'
        self.nombre = nuevo_nombre


# 3) Crear objeto
mi_objeto = MiClase()

#Test __str__()
print(str(mi_objeto))  # Imprime: mi clase

# 4) Testear __len__() mediante len()
print(len(mi_objeto))  # Imprime: 8 (largo de "mi clase")

# 5) Testear cambiar_nombre()
mi_objeto.cambiar_nombre("nuevo nombre")
print( str(mi_objeto))  # Imprime: nuevo nombre
print( len(mi_objeto))         # Imprime: 12 (largo de "nuevo nombre")
```

8.11.1. Múltiples gráficos en una figura (subplots)

```python
x = np . linspace (0 , 2* np . pi , 400)
y = np . sin ( x **2)

fig , axs = plt . subplots (2 , 2 , figsize =(10 ,8) )
axs [0 , 0]. plot (x , np . sin ( x ) )
axs [0 , 0]. set_title ('Seno ')

axs [0 , 1]. plot (x , np . cos ( x ) , color='tab:orange')
axs [0 , 1]. set_title (' Coseno ')

axs [1 , 0]. plot (x , y , color='tab:green')
axs [1 , 0]. set_title ( 'Seno de x ^2' )

axs [1 , 1]. plot (x , np . tan ( x ) ,color='tab:red')
axs [1 , 1]. set_ylim ([ -10 , 10]) # Limitar eje y
axs [1 , 1]. set_title ( 'Tangente')

plt . tight_layout () # Ajusta para que no se superpongan
plt . show ()
```
